```python
#File:  Case Study Using If_Else_While
#Author: Sophia Bracho
#Description: This short program asks for and accepts a student's first name and last name, asks for and accepts the student's
#gpa as a float, then it tests if the student's gpa is enough for the student to be in the Dean's List or The Honor Roll.
#the loop or the program will quit if "ZZZ" is entered as the first input.
while True:
    last_name=input("Enter student's last name: ['ZZZ to quit]")
    if last_name=="ZZZ":
        break
    first_name=input("Enter student's first name: ")
    student_gpa=input("Enter student's GPA: ")
    number=float(student_gpa)
    if number>=3.5:
        print(first_name,last_name,"has made it to the Dean's List!")
    if number >= 3.25:
        print(first_name,last_name,'has made it to the Honor Roll!')
    else:
        print("Don't give up! Keep studying and putting in the work!")
```
